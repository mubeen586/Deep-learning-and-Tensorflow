```python
from tensorflow.keras.preprocessing.image import ImageDataGenerator

train_datagen = ImageDataGenerator(rescale=1./255, validation_split=0.2)

train_generator = train_datagen.flow_from_directory(
    '../content/drive/My Drive/NumbersAndResnet50/numbers',  # Aapka dataset ka path
    target_size=(224, 224),  # ResNet50 ka input size
    batch_size=32,
    class_mode='categorical',
    subset='training')

validation_generator = train_datagen.flow_from_directory(
    '../content/drive/My Drive/NumbersAndResnet50/numbers', # Changed to the actual dataset path, same as training
    target_size=(224, 224),
    batch_size=32,
    class_mode='categorical',
    subset='validation') # This will use the validation_split specified in train_datagen
```

```python
from tensorflow.keras.applications import ResNet50
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, Flatten

model = Sequential()

resnet = ResNet50(include_top=False, weights='imagenet', input_shape=(224, 224, 3))

for layer in resnet.layers:
    layer.trainable = False  # Pretrained layers ko freeze karen

model.add(resnet)
model.add(Flatten())
model.add(Dense(512, activation='relu'))
model.add(Dense(10, activation='softmax'))  # 10 pages ke liye 10 classes

model.compile(optimizer='adam', loss='categorical_crossentropy', metrics=['accuracy'])
```

```python
history = model.fit(
    train_generator,
    epochs=10,  # 10 epochs, ya jitni zarurat ho
    validation_data=validation_generator
)
```
